```python
#Sieve
#==============================
def sieve(n):
  prime = [True] *(n+1)
  prime[0]=False
  prime[1]=True
  for i in range(2, n+1):
    if prime[i] == True:
      for j in range(i*2,n+1, i):
        prime[j] =False
  print("Prime Numbers Are: ")
  for i in range(n+1):
    if prime[i]:
      print(i, end=" ")
n = int(input("Enter A Number: "))
sieve(n)
#OUTPUT:
```

```python
#Ackermann
def ackermann(m, n):
    if m == 0:
        return n + 1
    elif n == 0:
        return ackermann(m - 1, 1)
    else:
        return ackermann(m - 1, ackermann(m, n - 1))
print(ackermann(2, 2))
#OUTPUT:
```

```python
#Anova Test
#==========
from scipy.stats import f_oneway
group1 = [80,85,78,90,88]
group2 = [75,70,72,80,74]
group3 = [90,92,88,91,89]

f_statistics,p_value = f_oneway(group1,group2,group3)
print("F-Statistic =",f_statistics)
print("P-value =",p_value)
if p_value<0.05:
  print("REJECT NULL HYPOTHESIS::")
else:
  print("ACCEPT NULL HYPOTHESIS::")

#OUTPUT
```

```python
#=================
#CHI-SQUARE
#================
import pandas as pd
from scipy.stats import chi2_contingency
data = [
    [18, 22, 10],   # Science Students
    [25, 15, 20]    # Arts Students
]

table = pd.DataFrame(
    data,
    columns=['Online Study', 'Classroom Study', 'Mixed Study'],
    index=['Science', 'Arts']
)

print("Contingency Table:")
print(table)
chi2, p_value, dof, expected = chi2_contingency(table)

print("\nChi-Square Statistic:", chi2)
print("Degrees of Freedom:", dof)
print("P-value:", p_value)

print("\nExpected Frequencies:")
print(pd.DataFrame(expected,
                   columns=table.columns,
                   index=table.index))
if p_value < 0.05:
    print("\nResult: Significant association between Student Type and Study Preference")
else:
    print("\nResult: No significant association between Student Type and Study Preference")

#OUTPUT
```

```python
#======================
#MUTUAL INFORMATION
#====================
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import mutual_info_score
df = pd.DataFrame({
    'Study_Habit': ['Regular','Regular','Irregular','Irregular','Regular','Irregular','Regular','Irregular'],
    'Result': ['Pass','Fail','Pass','Fail','Pass','Fail','Pass','Pass'],
    'Attendance': ['High','Low','High','Low','High','Low','High','Low']
})
columns = df.columns
mi_matrix = pd.DataFrame(0.0, index=columns, columns=columns)
for col1 in columns:
    for col2 in columns:
        mi_matrix.loc[col1, col2] = mutual_info_score(df[col1], df[col2])

print("Mutual Information Matrix:")
print(mi_matrix)

plt.figure(figsize=(6,5))
sns.heatmap(mi_matrix, annot=True, cmap="coolwarm")
plt.title("Mutual Information Heatmap")
plt.show()
#OUTPUT
```

```python
#=================
# McNemar Test
#================
from statsmodels.stats.contingency_tables import mcnemar
table= [
    [35, 5],
    [10, 50]
]
result = mcnemar(table, exact = False, correction = True)
print("McNemar Statistics:",result.statistic)
print("P-value:",result.pvalue)
if result.pvalue<0.05:
  print("Significant difference between paired observation::")
else:
  print("No significant difference between paired observations::")

#OUTPUT
```

```python
#=======================
# T-Test
#=====================
from scipy import stats
data = [23,25,21,22,24,26,27]
population_mean = 25
t_statistic, p_value = stats.ttest_1samp(data, population_mean)
print("ONE SAMPLE T-TEST RESULT")
print("-----------------------")
print("SAMPLE DATA: ",data)
print("POPULATION_MEAN :",population_mean)
print("T-Statistic:", t_statistic)
print("P-value:", p_value)

print("\nDECISION:")
if p_value < 0.05:
    print("Reject Null Hypothesis ,Significant Difference exists.")

else:
    print("FailTo Reject Null Hypothesis, NO Significant difference")

#OUTPUT:
```

```python
#=================
#Confusion Matrix

from sklearn.metrics import confusion_matrix,accuracy_score
from sklearn.metrics import matthews_corrcoef
y_true = [1, 0, 1, 1, 0, 1, 0, 1, 0, 1]
y_pred = [1, 0, 0, 1, 0, 1, 1, 1, 0, 1]
cm = confusion_matrix(y_true, y_pred)
print("Confusion Matrix:")
print(cm)
acc = accuracy_score(y_true, y_pred)
print("Accuracy:", acc)
mcc = matthews_corrcoef(y_true, y_pred)
print("MCC:", mcc)

#OUTPUT:
```

```python
#=======================
# Linear Regression
#=======================

import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
df = pd.DataFrame({
    'Study_Hours':[1,2,3,4,5],
    'Exercise_Hours':[0,1,1,2,3],
    'sleep_hours':[6,7,6,8,7],
    'performance_score':[40,50,55,70,85]
})
X = df[['Study_Hours','Exercise_Hours','sleep_hours']]
y = df['performance_score']
X = sm.add_constant(X)
model = sm.OLS(y,X).fit()
print(model.summary())

df['predicted_score'] = model.predict(X)
print("\nDataset With Prediction:")
print(df)
plt.scatter(df['Study_Hours'],df['performance_score'], color='blue')
plt.plot(df['Study_Hours'],df['predicted_score'],color='red')
plt.xlabel('Study Hours')
plt.ylabel('Performance Score')
plt.title("REGRESSION LINE(STUDY VS PERFORMANCE)")
plt.show()
#OUTPUT
```
